```python
import pandas as pd
df = pd.read_csv('YearDataDailyCollector.csv')
print(df.columns)
# print the days with the highest 'Michael - Emotion (5 neutral) ', 'Melanie - Emotion (5 neutral)' values - top 3
best = df.nlargest(10, ['Michael - Emotion (5 neutral) '])
for i,r in best[['Timestamp','Michael - Emotion (5 neutral) ','Best part of the day']].iterrows():
    print(r['Timestamp'],r['Michael - Emotion (5 neutral) '],r['Best part of the day'], sep='\t')
print()
best = df.nlargest(20, ['Melanie - Emotion (5 neutral)'])
for i,r in best[['Timestamp','Melanie - Emotion (5 neutral)','Best part of the day']].iterrows():
    print(r['Timestamp'],r['Melanie - Emotion (5 neutral)'],r['Best part of the day'], sep='\t')

sum(df['Melanie - Emotion (5 neutral)'] == 9)
```

```python
best = df.nsmallest(10, ['Michael - Emotion (5 neutral) '])
for i,r in best[['Timestamp','Michael - Emotion (5 neutral) ','Worst part of the day']].iterrows():
    print(r['Timestamp'],r['Michael - Emotion (5 neutral) '],r['Worst part of the day'], sep='\t')
print()
best = df.nsmallest(10, ['Melanie - Emotion (5 neutral)'])
for i,r in best[['Timestamp','Melanie - Emotion (5 neutral)','Worst part of the day']].iterrows():
    print(r['Timestamp'],r['Melanie - Emotion (5 neutral)'],r['Worst part of the day'], sep='\t')
```

```python
# read in this file: all_words_worst.txt and count the occurance of each word, one word per line

# create a dictionary where the key is the word and the value is the count
word_counts = {}
for line in open('all_words_best.txt'):
    word = line.strip()
    if word in word_counts:
        word_counts[word] += 1
    else:
        word_counts[word] = 1

# print the dictionary
for word, count in word_counts.items():
    print(word, count)
```
